```python
import pandas as pd

url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"

df = pd.read_csv(url)
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns

df.info()

df.describe()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
climate_df = df[['country',
                 'year',
                 'co2',
                 'population',
                 'gdp']]
```

```python
climate_df.head()
```

```python
climate_df.shape
```

```python
climate_df['co2']
```

```python
climate_df=climate_df.dropna()
```

```python
co2_yearly = climate_df.groupby('year')['co2'].sum()
```

```python
co2_yearly.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.plot(co2_yearly)

plt.xlabel("Year")
plt.ylabel("CO₂ Emissions")
plt.title("Global CO₂ Emissions Over Time")

plt.show()
```

```python
india = climate_df[climate_df['country'] == 'India']
```

```python
china = climate_df[climate_df['country'] == 'China']
```

```python
usa = climate_df[climate_df['country'] == 'United States']
```

```python
india.head()
```

```python
plt.figure(figsize=(12,6))

plt.plot(india['year'], india['co2'], label='India')
plt.plot(china['year'], china['co2'], label='China')
plt.plot(usa['year'], usa['co2'], label='USA')

plt.xlabel("Year")
plt.ylabel("CO₂ Emissions")
plt.title("CO₂ Emissions Comparison")

plt.legend()

plt.show()
```

```python
import plotly.express as py
```

```python
countries = climate_df[
    climate_df['country'].isin(
        ['India', 'China', 'United States']
    )
]
```

```python
fig = px.line(
    countries,
    x='year',
    y='co2',
    color='country',
    title='CO2 Emission Comparison'
)
fig.show()
```

Choropleth map colors regions according to data values.

```python
world_2020 = climate_df[climate_df['year'] == 2020]
```

```python
fig = px.choropleth(
    world_2020,
    locations='country',
    locationmode='country names',
    color='co2',
    hover_name='country',
    title='CO₂ Emissions by Country (2023)'
)

fig.show()
```

```python
top10 = world_2020.sort_values(
    by ='co2',
    ascending=False
).head(10)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
corr_df = climate_df[['co2','population','gdp']]
corr=corr_df.corr()
plt.figure(figsize=(8,5))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

Population vs CO₂

```python
fig = px.scatter(
    climate_df.dropna(),
    x='population',
    y='co2',
    color='country',
    title='Population vs CO₂'
)

fig.show()
```

```python
animated_df = climate_df[
    climate_df['year'] >= 1950
]

fig = px.choropleth(
    animated_df,
    locations='country',
    locationmode='country names',
    color='co2',
    hover_name='country',
    animation_frame='year',
    title='CO₂ Emissions Through Time'
)

fig.show()
```

```python
scatter_df = climate_df.dropna(
    subset=['gdp','co2']
)

fig = px.scatter(
    scatter_df,
    x='gdp',
    y='co2',
    color='country',
    hover_name='country',
    title='GDP vs CO₂'
)

fig.show()
```
